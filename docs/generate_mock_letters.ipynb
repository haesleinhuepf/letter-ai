# Generate mock letters

This notebook generates letters with randomish data. To ensure the letters fit to you and your domain, feel free to modify the following information and examples. 
The words provided here will enrich the random letter contents with actual content you specify.

```python
your_name = "Dr. Robert Haase"
your_institution = "Leipzig University"
your_cv = ["2023-?: Leipzig, Center for Scalable Data Analytics and Artificial Intelligence ScaDS.AI ",
           "2020-2023: TU Dresden, Cluster of Excellence Physics of Life",
           "2015-2020: Max Planck Institute for Molecular Cell Biology and Genetics (MPI-CBG)",
           "2008-2015: TU Dresden, OncoRay",
           "2006-2010: University of Applied Sciences HTW Dresden"]

random_first_names = ["Katharina", "Peter", "Luisa", "William", "Claudia", "John", "Maria", "Wen", "Stefanie", "Carlos"]
random_last_names = ["Wolf", "Jones", "Yi", "Haase", "Hoppenstedt", "Lindemann", "Hermmann", "Baum"]
job_profiles = ["Scientist", "Researcher", "Data Scientist", "Computer Scientist"]
domains = ["Medicine", "Clinical research", "Biological research", "Pharma Industry", "AI industry"]
degrees = ["Dr.", "Prof.", "MSc", "BSc", ""]
relationships = ["collaborator", "colleage", "former colleage", "student", "former student"]

folder = "letters/" # must end with /
```

```python
letter_types = ["support letter", 
                "recommendation letter",
                "letter of intent",
                "review report"]

num_letters = 5
```

```python
from openai import OpenAI
import numpy as np
import os
import subprocess
```

```python
def prompt_openai(prompt:str, model="gemma3:4b"):
    client = OpenAI(
        base_url="http://localhost:11434/v1",
        api_key="ollama"  # required but unused by Ollama
    )

    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content

prompt_openai("hello world")
```

```python
os.makedirs(folder, exist_ok=True)
```

```python
for letter_type in letter_types:
    for i in range(num_letters): 
        first_name = np.random.choice(random_first_names)
        last_name = np.random.choice(random_last_names)
        met_first_time_at = np.random.choice(your_cv)
        degree = np.random.choice(degrees)
        relationship = np.random.choice(relationships)
        job_profile = np.random.choice(job_profiles)
        domain = np.random.choice(domains)
    
        prompt = f"""
I want to write a {letter_type} for {degree} {first_name} {last_name}

The letter should outline that my {relationship} is an excellent {job_profile} working in {domain}.
We know each other from my time {met_first_time_at}.
The {letter_type} should be professional, positive and have a friendly tone. 
I really would like to promote {first_name} {last_name}.
Sign the letter with my name "{your_name}" and my institution "{your_institution}". 
Some up with details yourself. Do not add placeholders, but realistic names and institutions yourself.
The letter should be about one page long.

Leave out sender, addressee, additional explanations, todos. Just write the {letter_type} text including headline, greeting and closing.
"""
        letter = prompt_openai(prompt)
        print(letter[:100])
        print("-" * 60)
    
        filename = folder + letter_type.replace(" ", "_") + str(i) + ".txt"
        with open(filename, "w", encoding="utf-8") as f:
            f.write(letter)
        
```
